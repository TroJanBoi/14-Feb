## Cap the Outliners

***ถ้าอยาก ดู Outliners ให้ดูที่ Boxplot***

```python
df.boxplot(column="height") # Cap เข้ามาให้อยู่ในช่วงปกติ ดูอย่างนี้มันมี ส่วนสูง เกิน 400
```

```python
df.dropna().sort_values('height')[['height']].plot.bar(use_index=False, figsize=(20,3), xticks=[], xlim=(0,1200), ylim=(100,500))
```

```python
low, high = df["height"].quantile([0.05, 0.95])
print(low, high)
```

```python
df.loc[df[df["height"]<low].index,"height"]=low # ใครส่วนสูงต่ำกว่า low ที่เท่ากับ 151.786 ให้ค่าเท่ากับ low เลย
```

```python
df.loc[df[df["height"]>high].index,"height"]=high # ใครสูงเกินกว่า high ที่เท่ากับ 179.090 ให้ค่าเท่ากับ high เลย
```

```python
df.boxplot(column="height") # ค่า cap ให้เฉลี่ย เท่าๆ กัน
```

**Quiz !!**

> ลอง cap ที่ percentile 10-90

```python
# Answer

low, high = df["height"].quantile([0.1, 0.9])
print(low, high)
```

```python

df.loc[df[df["height"] < low ].index, "height"] = low
df.loc[df[df["height"] > high ].index, "height"] = high
```

```python
df.boxplot(column="height") #
```

## Bining คือการจัดกลุ่ม


```python
df["age"].hist()
```

```python
cut_labels = ['young', 'teen', 'adult', 'senior']
cut_bins = [0, 13, 25, 61, 100]
df['age_group'] = pd.cut(df['age'], bins=cut_bins, labels=cut_labels)
```

```python
df
```

```python
df.boxplot(column="age", by="age_group")
```

```python
df.boxplot(column="salary", by="age_group")
```

**Quiz !!**

> ทดลองทำ binning ของกลุ่มเงินเดือน Low (น้อยกว่า 50,000), Med (50,000 - 100,000), High (มากกว่า 100,000)

```python
# Answer
salary_labels = ['low', 'med', 'high']
salary_bins = [0, 50000, 100000, 10000000]
df["saraly_group"] = pd.cut(df["salary"], bins=salary_bins, labels=salary_labels)
```
